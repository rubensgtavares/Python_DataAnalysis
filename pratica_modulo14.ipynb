## 1\. Pacote Seaborn

**Seaborn** é um dos pacotes Python mais utilizados para visualização de dados. A documentação pode ser encontrada neste [link](https://seaborn.pydata.org/index.html). Possui diversas opções gráficos (barra, setores, linha, área, etc.) e uma excelente integração com os DataFrames do pacote Python Pandas. 

> O **Seaborn** usa o **Matplotlib** ([link](https://matplotlib.org/) da documentação), outro pacote bastante utilizada para visualização de dados.

### **1.1. Visualização** 

> O método do pacote Seaborn que constrói este gráfico é o `barplot` ([doc](https://seaborn.pydata.org/generated/seaborn.barplot.html)). 

```python
import seaborn as sns
import pandas as pd
```

```python
# O Seaborn tem uma série de datasets de exemplo pra ajudar no aprendizado
# Você consegue ver uma série de nomes de datasets já existente no Seaborn
# com o método .get_dataset_names()
sns.get_dataset_names()
```

```python
# Vamos trabalhar com esse dataset que configura uma situação fictícia 
# de um restaurante e seu dados gerados
data = sns.load_dataset('tips')
```

```python
data.head()
```

```python
# Estamos relacionando aqui os valores das compras com as gorgetas 
# Os valores estão relacionados por dia, o que é plotado no gráfico 
# São os dias da semana e os seus valores de contas e gorgetas
# Aparentimente não há datas específicas então não conseguir identificar
# Se cada ponto indica um dia completo ou uma compra completa, acredito
# que seja o segundo caso
sns.relplot(data=data, x='total_bill', y='tip', hue='day', style='day')
```

```python
# Podemos adicionar colunas ao gráfico, neste caso a coluna time tem somente dois valores
# Então é gerado dois gráficos relacionando esses dois valores, Dinner ou Lunch com os
# valores separados, acredito que se fossem mais valores seriam mais gráficos
sns.relplot(data=data, x='total_bill', y='tip', hue='day', style='day', col='time')
```

```python
# Pelo que eu entendi aqui podemos adicionar linhas e colunas de gráficos
# se você colocar uma série do dataset que tem 3 valores, no parâmetro das 
# colunas, teremos 3 gráficos um ao lado do outro, como colunas.
# Se colocarmos esses mesmos valores no parâmetros linha, será plotado 3 gráficos
# Um em cima do outro
sns.relplot(data=data, x='total_bill', y='tip', hue='day', style='day', col='time', row='sex')
```


## 2\. Categorias

### **2.1.  Gráfico de Barras** 

O **gráfico de barras** representa a relação entre uma variável categórica com uma variável numérica. Cada entidade da categoria é representada por uma barra, já a altura das barras representam os seus correspondente valor numérico. Útil para entender a distribuição de uma variável categórica.

> O método do pacote Seaborn que constrói este gráfico é o `barplot` ([doc](https://seaborn.pydata.org/generated/seaborn.barplot.html)). 

Algumas dicas:

 - Ordernar as barras pode gerar *insights*;
 - Barras horizontais podem facilitar a visualização.

Vamos utilizar a base de dados de **gorjetas**:

```python
data = sns.load_dataset("tips")
data.head()
```

```python
tips = data[["day","total_bill"]].groupby("day").agg("sum").reset_index()
tips.head()
```

```python
grafico = sns.barplot(data=tips, x="day",y="total_bill", ci=None, palette="dark")
grafico.set(title="Valor da conta por dia da semana", xlabel="Dia da semana", ylabel="Valor da conta (USD)");
```

```python
grafico = sns.barplot(data=tips, y="day",x="total_bill", errorbar=None, palette="dark")
grafico.set(title="Valor da conta por dia da semana", ylabel="Dia da semana", xlabel="Valor da conta (USD)");
```

```python
tips = data[["day","total_bill","time"]].groupby(["day","time"]).agg("sum").reset_index()
tips.head(20)
```

```python
grafico = sns.barplot(data=tips, x="day", y="total_bill", hue="time",palette="pastel")
grafico.set(title='Valor da conta por dia da semana por período', xlabel='Dia da Semana', ylabel='Valor da conta (USD)');
```

## Gráficos de Setores

Vamos lá, o professor fala que o gráfico de setores, o famoso gráfico de pizza, não é um bom gráfico pra analisar dados.
Ele é bom quando há uma desproporção muito considerável que dá pra ver no gráfico porque caso contrário não há uma visualização tão boa.
Cada entidade da categoria é representada por um setor de tamanho proporcional a sua respectiva proporção no todo.

Acredito que ele seja um bom detalhe estético em um dashboard mas sempre haverá a necessidade de complementar a análise com outro
gráfico com valores consistentes.

O pacote Seaborn não tem suporte para o gráfico de setores e recomenda o uso do gráfico de barras.


```python
# Não é recomendado mas e for usar;
# Se a proporção for em porcentagem, garanta que elas somem 100%;
#Procure colocar a legenda no gráfico;
# Não use gráficos 3D

import seaborn as sns
data = sns.load_dataset("tips")
data.head()
```

```python
tips = data[["tip","day"]].groupby("day").agg("sum").reset_index()
tips["tip_percent"] = 100 * tips["tip"] / tips["tip"].sum()
tips.head()
```

```python
tips.plot.pie(y="tip_percent", labels=tips["day"]);
```

### Tendências

#### Gráfico de linha
O gráfico de linha representa a evolução de uma variável numérica(eixo y), geralmente ao longo do tempo(eixo x), 
formando assim uma série temporal. Cada valor numérico é representado por pontos conectados por uma linha reta.

O método do pacote Seaborn que constrói este gráfico é o  `lineplot` ([doc](https://seaborn.pydata.org/generated/seaborn.lineplot.html)). 

Se uma das colunas for temporal(anos, meses, dias, horas, etc.) garanta a ordenação cronológica;
Muitas linhas em um mesmo gráfico por dificultar a visualização.


```python
# Vamos utilizar a base de dados de vôos

import seaborn as sns

data = sns.load_dataset("flights")
data.head()
```

```python
# Número de passageiros por ano:

flights =  data[["year", "passengers"]].groupby("year").agg("sum").reset_index()
flights.head()
```

```python
with sns.axes_style('whitegrid'):
    grafico = sns.lineplot(data=flights, x="year", y="passengers", palette="pastel")
    grafico.set(title='Passageiros por ano', xlabel='Ano', ylabel='Passageiros');
```

```python
# Número de passageiros por mês por ano:

flights = data 

with sns.axes_style('darkgrid'):
    grafico = sns.lineplot(data=data, x="month", y="passengers", hue="year", palette= "pastel")
    grafico.set(title='Passageiros por mês por ano', xlabel='Mês', ylabel='Passageiros')
    grafico.get_legend().set_title("Ano");
```

#### Gráfico de Área 

O **gráfico de área** é similar ao de **gráfico de linha** e representa a evolução de uma variável numérica (eixo `y`), geralmente ao longo do tempo (eixo `x`), formando assim uma `série temporal`.
Cada valor numérico é representado por pontos conectados por uma linha reta tendo ainda a área entre a linha e o eixo `x` preenchido por uma cor.

> O método do pacote Seaborn que constrói este gráfico é o `lineplot` ([doc](https://seaborn.pydata.org/generated/seaborn.lineplot.html)).

- Se uma das colunas for temporal (anos, meses, dias, horas, etc.) garanta a ordenação cronológica; 
- Muitas áreas em um mesmo gráfico por dificultar a visualização.

```python
# Vamos utilizar a base de dados de vôos

data = sns.load_dataset("flights")
data.head()
```

```python
flights = data[["year", "passengers"]].groupby("year").agg("sum").reset_index()
flights.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

with sns.axes_style("whitegrid"):

    grafico = sns.FacetGrid(data=flights, palette="pastel")
    grafico.map(sns.lineplot, "year","passengers")
    grafico.map(plt.fill_between, 'year', 'passengers', alpha=0.3)
    grafico.set(title='Passageiros por ano', xlabel='Ano', ylabel='Passageiros');
    grafico.fig.set_size_inches(w=7, h=4.5)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Configura o estilo do gráfico para um estilo pré-definido do Matplotlib chamado '_mpl-gallery'
plt.style.use('_mpl-gallery')

# Cria um array 'x' com 100 valores uniformemente espaçados entre 0 e 10
x = np.linspace(0, 10, 100)
# Gera os valores 'y' como uma função seno de 'x' multiplicada por 2, ajustada com um deslocamento vertical de 4
y = 4 + 1 * np.sin(2 * x)

# Cria um segundo array 'x2' com 25 valores uniformemente espaçados entre 0 e 10
x2 = np.linspace(0, 10, 25)
# Gera os valores 'y2' como uma função seno de 'x2' multiplicada por 2, ajustada com um deslocamento vertical de 4
y2 = 4 + 1 * np.sin(2 * x2)

# Cria uma figura e um conjunto de eixos (subplot)
fig, ax = plt.subplots(figsize=(20,12))

# Plota os valores 'x2' e 'y2' deslocados para cima em 2.5 unidades como marcadores 'x' com uma largura de borda de 2
ax.plot(x2, y2 + 2.5, 'x', markeredgewidth=4)

# Plota os valores 'x' e 'y' como uma linha contínua com largura de 2.0
ax.plot(x, y, linewidth=2.0)

# Plota os valores 'x2' e 'y2' deslocados para baixo em 2.5 unidades como pontos conectados por linhas com largura de 2
ax.plot(x2, y2 - 2.5, 'o-', linewidth=2)

# Define os limites dos eixos x e y, e define os ticks dos eixos x e y
ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 8), yticks=np.arange(1, 8))

# Exibe o gráfico na tela
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Dados
x = np.linspace(0, 2 * np.pi, 400)
y1 = np.sin(x)
y2 = np.sin(2 * x)

# Plotagem
plt.figure(figsize=(10, 5))

plt.plot(x, y1, label='sin(x)')
plt.plot(x, y2, label='sin(2x)', linestyle='--')

plt.legend()
plt.title('Comparação entre sin(x) e sin(2x)')
plt.xlabel('x')
plt.ylabel('y')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Definindo o estilo do gráfico
plt.style.use('_mpl-gallery')

# Gerando dados aleatórios
np.random.seed(3)  # Define a semente para o gerador de números aleatórios
x = 4 + np.random.normal(0, 2, 24)  # Array de 24 números aleatórios com média 4 e desvio padrão 2
y = 4 + np.random.normal(0, 2, len(x))  # Array de 24 números aleatórios com média 4 e desvio padrão 2

# Tamanhos e cores dos pontos
sizes = np.random.uniform(15, 80, len(x))  # Array de 24 números aleatórios entre 15 e 80 para os tamanhos dos pontos
colors = np.random.uniform(15, 80, len(x))  # Array de 24 números aleatórios entre 15 e 80 para as cores dos pontos

# Criando a figura e os eixos
fig, ax = plt.subplots()

# Plotando o gráfico de dispersão
ax.scatter(x, y, s=sizes, c=colors, vmin=0, vmax=100)  # Cria o gráfico de dispersão com os tamanhos e cores especificados

# Configurando os limites e os ticks dos eixos
ax.set(xlim=(0, 8), xticks=np.arange(1, 8),  # Limites do eixo x de 0 a 8 e ticks de 1 a 7
       ylim=(0, 8), yticks=np.arange(1, 8))  # Limites do eixo y de 0 a 8 e ticks de 1 a 7

# Exibindo o gráfico
plt.show()
```
